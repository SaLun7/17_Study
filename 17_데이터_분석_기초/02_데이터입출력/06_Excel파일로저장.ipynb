```python
import pandas as pd

data = {
    'name' : [ 'Jerry', 'Riah', 'Paul'],
    'algol' : [ "A", "A+", "B"],
    'basic' : [ "C", "B", "B+"],
    'c++' : [ "B+", "C", "C+"],
}

df = pd.DataFrame(data)
df = df.set_index("name")

print(df)
```

```python
df.to_excel("./data/df_sample.xlsx")
```

```python
df.to_excel("./data/df_sample_no_index.xlsx", index=False)
```

```python
import pandas as pd

data1 = {
    'name' : [ 'Jerry', 'Riah', 'Paul'],
    'algol' : [ "A", "A+", "B"],
    'basic' : [ "C", "B", "B+"],
    'c++' : [ "B+", "C", "C+"],
}

data2 = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11 ,12],
    "c4" : [13, 14, 15]
}

df1 = pd.DataFrame(data1)
df1 = df1.set_index("name")

print(df1)
```

```python
df2 = pd.DataFrame(data2)
df2 = df2.set_index("c0")

print(df2)
```

```python
with pd.ExcelWriter("./data/df_excelwriter.xlsx") as writer:
    df1.to_excel(writer, sheet_name="sheet1")
    df2.to_excel(writer, sheet_name="sheet2")
```
