```python
# 데이터 개수 확인

df = pd.read_csv("./data/auto-mpg.csv", header=None)
print(df)

df.columns = [
    "mpg", "cylinders", "displacement", "horsepower", "weight", "acceleration", "model year", "origin", "name"
]

print(df)
```

```python
print(df.count())
```

```python
print(type(df.count()))
```

```python
# 각 열의 고유값 개수

# DataFrame객체["열이름"].value_counts()

unique_value = df["origin"].value_counts()
print(unique_value)

unique_value = df["origin"].value_counts().iloc[1]
print(unique_value)
```

```python
unique_values_ratio = df["origin"].value_counts(True)
print(unique_values_ratio)
```

```python
(df["origin"].value_counts(True) * 100).round(1)
```
