```python
# 열분리
import pandas as pd
df = pd.read_excel("./data/주가데이터.xlsx", engine="openpyxl")
df.head()
```

```python
df.info()
```

```python
df['연월일'] = df['연월일'].astype('str')   
dates = df['연월일'].str.split('-')        

dates
```

```python
# Series객체.str.get(인덱스)
# Series객체.str[인덱스]

df['연'] = dates.str.get(0)     
# df["연"].head()
df['월'] = dates.str.get(1)  
# df["월"].head()  
df['일'] = dates.str.get(2)
# df["일"].head()
```
