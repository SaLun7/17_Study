```python
import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "sex", "class", "fare", "survived"]]
df.head()
```

```python
df2 = pd.pivot_table(
    df,
    index=["class", "survived"],
    columns="sex",
    values="age",
    aggfunc="mean",
    observed=True
)

df2
```

```python
df_stacked = df2.stack()
df_stacked
```

```python
df_unstacked = df2.unstack()
df_unstacked
```

```python
df2
```

```python
df2.unstack(level=0)
```

```python
df2.unstack(level=1)
```

```python
# DataFrame객체.melt(id_vars=[], vars="")

import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["John", "Mary"],
    "City": ["New York", "Los Angeles"],
    "Age": [28, 32],
    "Salary": [50000, 60000],
})

df
```

```python
df_melted = df.melt(
    id_vars=["Name", "City"], 
    var_name="Attribute", 
    value_name="Value"
)

df_melted
```
