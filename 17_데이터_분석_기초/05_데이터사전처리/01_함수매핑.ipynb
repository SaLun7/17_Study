```python
# 데이터프레임객체.apply(매핑함수, axis=0)

df.head()
```

```python
def calculate_stats(col):
    max_val = col.max()
    min_val = col.min()
    mean_val = col.mean()
    median_val = col.median()

    df = pd.Series([
        max_val, min_val, mean_val, median_val
    ], index=["Max", "Min", "Mean", "Median"])
    return df


result_df = df.apply(calculate_stats, axis=0)
result_df
```

```python
result_sr = df.apply(lambda x : x.max() - x.min(), axis=0)

result_sr
```

```python
df.head(2)
```

```python
# 데이터프레임객체.apply(매핑함수, axis=1)

def calulate_diff_avg(row):
    diff = row.max() - row.min()
    avg = row.mean()

    return pd.Series([diff, avg], index = ["차이", "평균"])

result_df2 = df.apply(calulate_diff_avg, axis=1)
result_df2
```

```python
result_df3 = df.apply(
    lambda row, multiplier : pd.Series([(row.max() - row.min()) * multiplier, row.mean()],
    index=["차이", "평균"]),
    multiplier = 2, axis = 1
)

result_df3
```

```python
# 데이터프레임 객체에 함수 매핑
# DataFrame객체.pipe(매핑함수)

df1 = titanic.loc[:, ["age", "fare"]]
df2 = titanic.loc[:, ["embark_town", "embarked"]]

def missing_value(df):
    return df.isnull()

result_df = df1.pipe(missing_value)
result_df
```

```python
def missing_count(df):
    return missing_value(df).sum()

result_series = df1.pipe(missing_count)
result_series
```

```python
def totoal_number_missing(df):
    return missing_count(df).sum()

result_value = df1.pipe(totoal_number_missing).sum()
result_value
```

```python
def extract_initial(df):
    df["town_initial"] = df["embark_town"].str[0]
    return df

def verify_initial(df):
    df["verified"] = df["embarked"] == df["town_initial"]
    return df

verify_initial(extract_initial(df2))
```
