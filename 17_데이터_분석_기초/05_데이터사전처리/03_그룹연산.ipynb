```python
# 그룹객체만들기
# DataFrame객체.groupby(기준이되는열)

import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "sex", "class", "fare", "survived"]]
df.head()
```

```python
df["class"].value_counts()
df["class"].nunique()
df["class"].unique()
```

```python
grouped = df.groupby("class", observed=True)
grouped
```

```python
for key, group in grouped:
    print("* kwt * : ", key)
    print("* number * : ", len(group))
    print(group.head())
```

```python
average = grouped.mean(numeric_only=True)
average
```

```python
# 개별그룹 선택하기
group3 = grouped.get_group("Third")
group3
```

```python
group1 = grouped.get_group("First")
group1
```

```python
# 여러열을 기준으로 그룹화
# DataFrame객체.groupby(기준이되는열의리스트)

df.head()
```

```python
# aggregate() 매서드 데이터 집계: group객체.aggregate(매핑함수)
# agg() 매서드 데이터 집계: group객체.agg(매핑함수)

agg_mean = grouped.aggregate("mean", numeric_only = True)
agg_mean
```

```python
grouped.agg("mean", numeric_only=True)
```

```python
# group객체.agg([함수1, 함수2])
# group객체.agg(["열1": 함수1, "열2": 함수2])

agg_all = grouped.agg(["min", "max"])
agg_all
```

```python
agg_sep = grouped.agg({
    "fare": ["min", "max"],
    "age": "mean"
})

agg_sep
```

```python
def min_max(x):
    return x.max() - x.min()

agg_minmax = grouped[["age", "fare"]].agg(min_max)
agg_minmax
```

```python
# 변환  Transformation

titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "sex", "class", "fare", "survived"]]
df.head()
```

```python
grouped = df.groupby(["class"], observed=True)
grouped
```

```python
grouped["fare"].cumsum()
```

```python
df["fare_cumsum"] = grouped["fare"].cumsum()
df.head(10)
```

```python
# grouped객체.transform(매핑함수)

grouped[["fare"]].transform("cumsum")
```

```python
grouped[["age", "survived"]].transform("mean")
```

```python
def z_score(x):
    return (x -x.mean()) / x.std()

age_zscore = grouped["age"].transform(z_score)
age_zscore
```

```python
age_zscore2 = grouped["age"].transform(lambda x : (x-x.mean())/x.std())
age_zscore2
```
