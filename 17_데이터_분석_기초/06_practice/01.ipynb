## 문제 ② Min-Max / Z 정규화 평균 비교

`Wind` 변수에 대해서 **Min-Max 정규화**를 수행한 후 평균값과  
**Z 정규화(표준화)**를 수행한 후 평균값의 차이를 구하시오.

- Min-Max 정규화: $(x - min) / (max - min)$
- Z 정규화: $(x - mean) / std$

> 출력: 두 평균의 차이

````python
# 정답
airquality["Wind"].head()
wind_min = airquality["Wind"].min()
wind_max = airquality["Wind"].max()
wind_mean = airquality["Wind"].mean()
wind_std = airquality["Wind"].std()

airquality["minmax"] = (airquality["Wind"] - wind_min) / wind_max - wind_min

airquality["zscor"] = (airquality["Wind"] - wind_mean) / wind_std

result2 = airquality["minmax"].mean() - airquality["zscor"].mean()
result2 = round(result2, 2)
print(result2)
````

````python

def mm(a):
    return (a-airquality["Wind"].min())/(airquality["Wind"].max() - airquality["Wind"].min())
    

res = airquality["Wind"].apply(mm).mean()
# print(res)

def z(b):
    return (b-airquality["Wind"].mean()) / airquality["Wind"].std()

res2 = airquality["Wind"].apply(z).mean()
# print(res2)

result2 = res - res2
print(round(result2, 2))
````
