```python
import pandas as pd

train = pd.read_csv("./data/bike_train.csv")
test = pd.read_csv("./data/bike_test.csv")

print(train.head())
```

```python
print(train.info())
```

```python
print(test.info())
```

```python
del train["casual"]
del train["registered"]
```

```python
train["datetime"] = pd.to_datetime(train["datetime"])
test["datetime"] = pd.to_datetime(test["datetime"])
print(train.info())
```

```python
train["week"] = train["datetime"].dt.weekday
train["hour"] = train["datetime"].dt.hour
train["year"] = train["datetime"].dt.year
train["month"] = train["datetime"].dt.month
train["day"] = train["datetime"].dt.day

display(train.head())
```

```python
test["week"] = test["datetime"].dt.weekday
test["hour"] = test["datetime"].dt.hour
test["year"] = test["datetime"].dt.year
test["month"] = test["datetime"].dt.month
test["day"] = test["datetime"].dt.day
```

```python
del train["datetime"]
del test["datetime"]
```

```python
X = train.drop("count", axis=1)
Y = train["count"]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure()
sns.boxplot(y=train["count"])
plt.title("count boxplot")
plt.ylabel("count")
plt.show()
```

```python
import numpy as np

Y_log = np.log1p(Y)
```

```python
from sklearn.model_selection import train_test_split

train_x, valid_x, train_y, valid_y = train_test_split(X, Y_log, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(train_x, train_y)
print(model.score(train_x, train_y))
print(model.score(valid_x, valid_y))
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

def print_score(valid_x, valid_y):
    pred = model.predict(valid_x)
    print(f"MAE: {mean_absolute_error(valid_y, pred)}")
    print(f"RMSE: {np.sqrt(mean_squared_error(valid_y, pred))}")
    print(f"R2_SCORE: {r2_score(valid_y, pred)}")

print_score(valid_x, valid_y)
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state = 42, max_depth = 4)
model.fit(train_x, train_y)
print(model.score(train_x, train_y))
print(model.score(valid_x, valid_y))
```

```python
print_score(valid_x, valid_y)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(train_x, train_y)
print(model.score(train_x, train_y))
print(model.score(valid_x, valid_y))
```

```python
print_score(valid_x, valid_y)
```

```python
from xgboost import XGBRegressor

model = XGBRegressor()
model.fit(train_x, train_y)
print(model.score(train_x, train_y))
print(model.score(valid_x, valid_y))
```

```python
print_score(valid_x, valid_y)
```

```python
pred = model.predict(test)
pred = np.expm1(pred).round(0)

print(pred)
```
