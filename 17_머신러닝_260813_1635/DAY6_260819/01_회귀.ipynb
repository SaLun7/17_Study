```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("./data/house_train.csv")
df.head()
```

```python
df.info()
```

```python
tmp = df.isnull().sum()
print(tmp[tmp>0].sort_values(ascending=False))
```

```python
plt.xticks(rotation=15)
sns.histplot(df['SalePrice'])
plt.show()
```

```python
log_SalePrice = np.log1p(df['SalePrice'])
sns.histplot(log_SalePrice)
plt.show()
```

```python
df["FireplaceQu"].value_counts()
```

```python
drop_cols = ['Id', 'PoolQC' , 'MiscFeature', 'Alley', 'Fence','FireplaceQu', 'MasVnrType']
```

```python
df.drop(drop_cols, axis = 1, inplace=True)
num_cols = df.dtypes[df.dtypes != 'str'].index
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0].sort_values(ascending = False))
```

```python
from sklearn.model_selection import train_test_split

y = df["SalePrice"]
X = df.drop(["SalePrice"], axis = 1)

print(y.shape)
print(X.shape)

X_dum = pd.get_dummies(X)

X_train, X_val, y_train, y_val = train_test_split(X_dum, y, test_size = 0.2, random_state = 42)
```

```python
test = pd.read_csv("./data/house_test.csv")

test.head()
```

```python
test.info()
```

```python
tmp1 = test.isnull().sum()
print(tmp1[tmp1>0].sort_values(ascending=False))
```

```python
test_id = test["Id"]

test.drop(drop_cols, axis = 1, inplace=True)
num_cols = test.dtypes[test.dtypes != 'str'].index
test[num_cols] = test[num_cols].fillna(test[num_cols].mean())
```

```python
print(test.isnull().sum()[test.isnull().sum() > 0].sort_values(ascending = False))
```

```python
print(df.shape)
print(test.shape)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

xgb = XGBRegressor()
xgb.fit(X_train, y_train)
print(xgb.score(X_train, y_train))
print(xgb.score(X_val, y_val))

pred = xgb.predict(X_val)
print(f"MAE: {mean_absolute_error(y_val, pred)}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_val, pred))}")
print(f"R2_SCORE: {r2_score(y_val, pred)}")
```

```python
from lightgbm import LGBMRegressor

lgb = LGBMRegressor()
lgb.fit(X_train, y_train)
print(lgb.score(X_train, y_train))
print(lgb.score(X_val, y_val))

pred1 = lgb.predict(X_val)
print(f"MAE: {mean_absolute_error(y_val, pred1)}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_val, pred1))}")
print(f"R2_SCORE: {r2_score(y_val, pred1)}")
```

```python
test_dum = pd.get_dummies(test)

test_dum = test_dum.reindex(columns=X_dum.columns,fill_value=0)

xgb_pred = xgb.predict(test_dum)
xgb_pred = xgb_pred.round(0)

print(xgb_pred)
```

```python
lgb_pred = lgb.predict(test_dum)
lgb_pred = lgb_pred.round(0)

print(lgb_pred)
```

```python
result = pd.DataFrame({
    "Id": test_id,
    "XGB_SalePrice": xgb_pred,
    "LGB_SalePrice": lgb_pred
})

result.to_csv("result.csv", index=False)
```
