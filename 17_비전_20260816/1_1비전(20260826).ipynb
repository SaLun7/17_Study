```python
# gpt 돌린 거
import matplotlib.pyplot as plt

if image is None:
    raise FileNotFoundError("like_lenna.png 파일을 읽을 수 없습니다.")

plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```

```python
print(image.shape)
```

```python
# resize
image_small = cv2.resize(image, (100, 100))
plt.imshow(image_small, cmap= "gray")
plt.title("SMALL Like Lenna")
plt.axis("off")
plt.show()
```

```python
image_flip = cv2.flip(image, 0)

plt.imshow(image_flip, cmap= "gray")
plt.title("Flip Like Lenna")
plt.axis("off")
plt.show()
```

```python
image_flip = cv2.flip(image, 1)

plt.imshow(image_flip, cmap= "gray")
plt.title("Flip Like Lenna")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape
matrix = cv2.getRotationMatrix2D((width / 2, height / 2), 90, 1)
res = cv2.warpAffine(image, matrix, (width, height))

plt.imshow(res, cmap= "gray")
plt.title("Rotation Like Lenna")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape
matrix = cv2.getRotationMatrix2D((width / 2, height / 2), 30, 1)
res = cv2.warpAffine(image, matrix, (width, height), borderValue= 0)

plt.imshow(res, cmap= "gray")
plt.title("30 Rotation Like Lenna")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape
matrix = cv2.getRotationMatrix2D((width / 2, height / 2), 30, 1)
res = cv2.warpAffine(image, matrix, (width, height), borderValue=255)

plt.imshow(res, cmap= "gray")
plt.title("30 Rotation Like Lenna")
plt.axis("off")
plt.show()
```
