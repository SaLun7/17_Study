```python
# 머신러닝에 필요한 여러 기능을 제공하는 scikit-learn 라이브러리를 불러온다
import sklearn
# sklearn의 예제 데이터인 iris 데이터를 불러오는 load_iris 함수를 가져온다
from sklearn.datasets import load_iris
# 결정 트리를 이용해 분류 모델을 학습하고 예측하기 위해 가져온다
from sklearn.tree import DecisionTreeClassifier
# 데이터를 모두 학습용과 성능 평과용으로 나누기 위해 가져온다
from sklearn.model_selection import train_test_split
# 표 형태의 데이터를 다루고 분석하기 위해 pandas를 pd라는 이름으로 불러온다
import pandas as pd
```

```python
print(load_iris())
```

```python
# load_iris() 함수를 실행해 iris 데이터셋을 불러오고 iris 변수에 저장한다
iris = load_iris()

# iris 데이터에서 꽃의 특징(피처) 데이터를 가져와 iris_data에 저장한다
iris_data = iris.data

# iris 데이터에서 모델이 예측해야 할 꽃의 종류(정답)를 iris_label에 저장한다
iris_label = iris.target

# print(iris_label)
# print(iris.target_names)

# iris_df = pd.DataFrame(data = iris_data, columns=iris.feature_names)
# iris_df["label"] = iris.target
# iris_df.tail()
```

```python
# iris의 X와 y를 학습용 80%, 테스트용 20%로 나눈다
# random_state를 지정해 실행할 때마다 동일하게 데이터가 나뉘도록 한다
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label, 
                                                    test_size=0.2, 
                                                    random_state=42)
```

```python
# 결정 트리 분류 모델을 생성하고, 같은 조건에서 결과를 재현할 수 있도록 random_state를 42로 고정한다
dt_clf = DecisionTreeClassifier(random_state=42)

# 학습용 문제 X_train과 정답 y_train을 이용해 결정 트리 모델을 학습시킨다
dt_clf.fit(X_train, y_train);
```

```python
# 테스트용 문제 X_test를 학습된 결정 트리 모델에 넣어 예측하고, 그 예측값을 pred에 저장한다
pred = dt_clf.predict(X_test)

# 예측 값을 출력한다
print(pred)
```

```python
# 실제 정답과 예측값을 비교해 정확도를 계산하기 위해 accuracy_score를 불러온다
from sklearn.metrics import accuracy_score

# 실제 정답 y_test와 모델의 예측값 pred를 비교해 정확도를 출력한다
print(accuracy_score(y_test, pred))
```
