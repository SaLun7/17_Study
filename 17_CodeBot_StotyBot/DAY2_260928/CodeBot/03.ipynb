```python
import pickle
from pathlib import Path
from CodeBot.tokenizer import train_bpe

codebot_dir = Path('CodeBot')

vocab_size = 1000

text = (codebot_dir / 'tiny_codes.txt').read_text(encoding='utf-8')
merge_rules = train_bpe(text, vocab_size)

with (codebot_dir / 'merge_rules.pkl').open('wb') as f:
    pickle.dump(merge_rules, f)

print(f'merge_rules.pkl 저장 완료: {len(merge_rules)}개 (vocab_size={vocab_size})')
```
