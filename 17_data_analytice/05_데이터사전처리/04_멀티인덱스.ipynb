```python
import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "sex", "class", "fare", "survived"]]
df.head()
```

```python
grouped = df.groupby(["class", "sex"], observed=True)
grouped
```

```python
gdf = grouped.agg(["mean", "std"], numeric_only = True)
gdf
```

```python
gdf.index
```

```python
arrays = [
    ["a", "a", "b", "b"],
    [1, 2, 1, 2]
]

multi_index_array = pd.MultiIndex.from_arrays(arrays, names=("letter", "number"))
multi_index_array
```

```python
pd.DataFrame([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
], index=multi_index_array)
```

```python
tuples = [
    ("a", 1),
    ("a", 2),
    ("b", 1),
    ("b", 2),
]

multi_index_tuple = pd.MultiIndex.from_tuples(tuples, names=("letter", "number"))
multi_index_tuple
```

```python
df = pd.DataFrame(
    [
        ["a", 1],
        ["a", 2],
        ["b", 1],
        ["b", 2],
    ],
    columns=["letter", "number"]
)

df
```

```python
multi_index_frame = pd.MultiIndex.from_frame(df)
multi_index_frame
```

```python
multi_index_frame.get_level_values(1)
```

```python
multi_index_frame.get_level_values("letter")
```
