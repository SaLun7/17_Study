```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
import matplotlib.pyplot as plt

sns.set_style("darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(
    x = "age",
    y = "fare",
    data=titanic,
    ax=axes[0]
)

sns.regplot(
    x = "age",
    y = "fare",
    data=titanic,
    ax=axes[1],
    fit_reg=False
)


plt.show()
```

```python
# 히스토그램/커널 밀도 그래프

fig, axes = plt.subplot_mosaic([
    ["top_left", "top_right"],
    ["middle_left", "middle_right"],
    ["bottom", "bottom"]
])

sns.histplot(
    x="age",
    data=titanic,
    ax=axes["top_left"]
)

# 2. hue옵션 추가
sns.histplot(
    x="age",
    hue="survived",
    data=titanic,
    ax=axes["top_right"]
)

# dodge 적용
sns.histplot(
    x="age",
    hue="survived",
    data=titanic,
    ax=axes["middle_left"],
    multiple="dodge"
)

# stack
sns.histplot(
    x="age",
    hue="survived",
    multiple="stack",
    data=titanic,
    ax=axes["middle_right"]
)

# multiple = fill
sns.histplot(
    x="age",
    hue="survived",
    multiple="fill",
    data=titanic,
    ax=axes["bottom"]
)

plt.show()
```

```python
fig, axes = plt.subplot_mosaic([
    ["top_left", "top_center", "right"],
    ["bottom_left", "bottom_center", "right"]
], figsize=(15, 6), constrained_layout=True)

sns.kdeplot(x="age", data=titanic, ax=axes["top_left"])

sns.kdeplot(x="age", hue="survived", data=titanic, ax=axes["bottom_left"])

sns.kdeplot(x="age", hue="survived", data=titanic, fill=True, ax=axes["top_center"])

sns.kdeplot(
    x="age",
    hue="survived",
    multiple="stack",
    data=titanic,
    ax=axes["bottom_center"]
)

sns.kdeplot(
    x="age",
    hue="survived",
    multiple="fill",
    bw_adjust=2.0,
    data=titanic,
    ax=axes["right"]
)

plt.show()
```

```python
# 히트맵

sns.set_style("darkgrid")

table = titanic.pivot_table(
    index=["sex"],
    columns=["class"],
    aggfunc="size",
    observed=False
)

table
```

```python
sns.heatmap(
    table,
    annot=True, fmt="d",
    linewidths=5,
    cbar=True
)

plt.show()
```

```python
# 범주형 데이터의 산점도

titanic.head()
```

```python
sns.set_style("whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.stripplot(
    x="class",
    y="age",
    data=titanic,
    ax=axes[0]
)

sns.stripplot(
    x="class",
    y="age",
    data=titanic,
    ax=axes[1],
    hue="class",
    size=4
)
```

```python
# 막대그래프

sns.set_style("ticks")

fig, axes = plt.subplots(2, 2, figsize=(10, 6), constrained_layout=True)

sns.barplot(
    x="sex",
    y="survived",
    data=titanic,
    ax=axes[0,0]
)

sns.barplot(
    x="sex",
    y="survived",
    data=titanic,
    errorbar=None,
    ax=axes[0,1]
)

sns.barplot(
    x="sex",
    y="survived",
    hue="class",
    data=titanic,
    errorbar=("ci", 95),
    ax=axes[1,0],
    estimator="median"
)

sns.barplot(
    x="sex",
    y="survived",
    hue="class",
    dodge=False,
    data=titanic,
    ax=axes[1,1]
)

plt.show()
```

```python
# 박스플롯 / 바이올린 그랲

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.boxplot(x='alive', y='age', data=titanic, ax=axes[0,0]) 

sns.boxplot(x='alive', y='age', hue='sex', data=titanic, ax=axes[0,1]) 

sns.violinplot(x='alive', y='age', data=titanic, ax=axes[1,0]) 

sns.violinplot(x='alive', y='age', hue='sex', data=titanic, ax=axes[1,1]) 
```
