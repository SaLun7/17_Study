## 문제 ① 구급대 평균 이동시간

주어진 데이터를 이용하여 **이동시간**(출동시간과 도착시간의 차이)이  
평균적으로 **가장 오래 걸린 구급대**의 이동시간을 구하시오.

- 분 단위로 환산
- 반올림하여 **정수**로 출력

**데이터**: `data_6_1_1.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_1.csv
```

> 출력: 평균 이동시간(분, 정수)

````python
df["출동시간"] = pd.to_datetime(df["출동시간"])
df["도착시간"] = pd.to_datetime(df["도착시간"])
df.info()
````

````python
df["출동시간_분"] = df["출동시간"].dt.minute
df["도착시간_분"] = df["도착시간"].dt.minute
df.head()
````

````python
df["차이"] = df["도착시간_분"] - df["출동시간_분"]
result1 = df.groupby("구급대")["차이"].mean().max()
print(round(result1, 2))
````

## 문제 ② 교수 1인당 학생 수가 가장 많은 대학

주어진 데이터를 이용하여 **교수 1인당 맡고 있는 학생 수**가 가장 많은 대학교를 찾고,  
그 대학교의 **전체 교수 수**를 구하시오. (**정수**로 출력)

**데이터**: `data_6_1_2.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv
```

> 출력: 해당 대학교의 전체 교수 수(정수)

````python
sum_stu = df2.loc[:, "경영학과" : "영문학과"].sum(axis=1)
sum_stu.head()
````

````python
df2["교수1인당학생수"] = sum_stu / df2["교수"]
df2.head()
````

````python
df2 = df2.sort_values("교수1인당학생수", ascending=False)
df2.head()
````
