```python
# 공용 자료와 모듈 경로를 현재 저장소에서 찾는다.
from pathlib import Path
import os, sys

_course_name = "17_CodeBot_StoryBot"
_start = Path.cwd().resolve()
_candidates = [p for base in (_start, *_start.parents) for p in (base, base / _course_name)]
PROJECT_ROOT = next((p for p in _candidates if p.name == _course_name and p.is_dir()), None)
if PROJECT_ROOT is None:
    raise FileNotFoundError(f"17_Study 안에서 실행해 주세요: {_course_name}")
os.chdir(PROJECT_ROOT / "DAY3_260929")
sys.path.insert(0, str(PROJECT_ROOT))

sys.path.insert(0, str(Path.cwd()))

import pickle
from pathlib import Path
from CodeBot.tokenizer import train_bpe

codebot_dir = Path('CodeBot')

vocab_size = 1000

text = (codebot_dir / 'tiny_codes.txt').read_text(encoding='utf-8')
merge_rules = train_bpe(text, vocab_size)

with (codebot_dir / 'merge_rules.pkl').open('wb') as f:
    pickle.dump(merge_rules, f)

print(f'merge_rules.pkl 저장 완료: {len(merge_rules)}개 (vocab_size={vocab_size})')
```
